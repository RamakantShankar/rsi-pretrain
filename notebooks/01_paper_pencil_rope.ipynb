## 1. The 2D Rotation Formula

Given an initial 2D vector $[x_1, \; x_2]$ and a rotation angle $\theta$:

$$
x'_1 = x_1 \cos\theta - x_2 \sin\theta
$$
$$
x'_2 = x_1 \sin\theta + x_2 \cos\theta
$$

*(On a clock face, this rotates the arrow by angle $\theta$ around the origin)*.
