## 3. Exercise 1: The Automated Triage Algorithm on Paper 📝

Suppose the agent monitors a training run and reads the last 3 steps from <code>metrics.jsonl</code>:
<pre style='font-size: 14px; background: #161b22; color: #c9d1d9; padding: 12px; border-radius: 8px;'>
- Step 110: Loss = 2.10, GradNorm = 0.85
- Step 120: Loss = 2.05, GradNorm = 0.90
- Step 130: Loss = 6.40, GradNorm = 8.20   (Sudden rogue batch!)
</pre>

### Step-by-Step on Paper:
1. <b>Calculate the moving average loss before the anomaly:</b><br>
   Moving Avg = (2.10 + 2.05) / 2 = <b>2.075</b>
<br><br>
2. <b>Calculate the Spike Ratio:</b><br>
   Spike Ratio = Current_Loss / Moving_Avg = 6.40 / 2.075 ≈ <b>3.08</b>
<br><br>
3. <b>Agent Decision Rule:</b><br>
   - Is Spike Ratio > 2.5? <b>YES! (Anomaly Alert!)</b><br>
   - <i>Action 1:</i> Terminate runaway training job immediately.<br>
   - <i>Action 2:</i> Load clean checkpoint from Step 120 (<code>best_model.pt</code>).<br>
   - <i>Action 3:</i> Reduce learning rate by 30%: New LR = 0.70 · Old LR.<br>
   - <i>Action 4:</i> Resume training automatically!

<br>
<b>Result:</b> The training run self-heals overnight with zero human intervention required!
