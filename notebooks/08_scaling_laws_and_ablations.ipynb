<b>The Chinchilla Discovery (DeepMind, 2022):</b><br>
Parameters (N) and Tokens (D) should be scaled in <b>equal 1:1 proportion</b>:<br>
<pre style='font-size: 16px; background: #161b22; color: #3fb950; padding: 12px; border-radius: 8px;'>
Compute-Optimal Ratio:  Tokens (D) ≈ 20 × Parameters (N)
</pre>

## 3. Exercise 1: Sizing a Model for Given Compute on Paper 📝

The total floating-point operations (FLOPs) required to train a Transformer is approximately:
<pre style='font-size: 15px; background: #161b22; color: #58a6ff; padding: 12px; border-radius: 8px;'>
Compute C ≈ 6 · N · D
(2·N FLOPs for forward pass + 4·N FLOPs for backward pass per token)
</pre>

Suppose our DGX GPU cluster gives us a compute budget of <b>C = 1.2 · 10¹⁷ FLOPs</b>.<br>
Using the Chinchilla optimal rule: <b>D = 20 · N</b>.

### Step-by-Step on Paper:
1. <b>Substitute D into the compute equation:</b><br>
   C = 6 · N · (20 · N) = <b>120 · N²</b>
<br><br>
2. <b>Solve for N²:</b><br>
   N² = (1.2 · 10¹⁷) / 120 = <b>1.0 · 10¹⁵</b>
<br><br>
3. <b>Take the square root to find optimal model size N:</b><br>
   N = sqrt( 1.0 · 10¹⁵ ) ≈ <b>31.6 Million Parameters</b>
<br><br>
4. <b>Calculate optimal training dataset size D:</b><br>
   D = 20 · 31.6 Million ≈ <b>632 Million Tokens</b>

<br>
<b>The Scientific Takeaway:</b> Before training begins, you know exactly that building a 31.6M parameter model and preparing 632M tokens gives you the absolute lowest loss for your compute budget!

## 4. Verify Chinchilla Compute Calculation in Python

Run this cell to verify our compute calculations and print scaling ratios:

```python
import math

compute_flops = 1.2e17

# Chinchilla constant: C = 120 * N^2
optimal_params = math.sqrt(compute_flops / 120.0)
optimal_tokens = 20.0 * optimal_params

print(f"Compute Budget:       {compute_flops:.1e} FLOPs")
print(f"Optimal Parameters N: {optimal_params / 1e6:.1f} Million parameters")
print(f"Optimal Tokens D:     {optimal_tokens / 1e6:.1f} Million tokens")
print(
    f"Verify FLOPs:         {6.0 * optimal_params * optimal_tokens:.1e} FLOPs (Exact match!)"
)
```
